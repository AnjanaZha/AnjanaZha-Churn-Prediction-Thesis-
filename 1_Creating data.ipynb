```python
import numpy as np
import pandas as pd
```

```python
# Define the number of rows for synthetic data (10 million)
num_rows = 10_000_000


```

```python
# Generate synthetic data based on the structure of Sheet2
num_rows = 5000000

def generate_synthetic_data(txn_flow,min_product_id=10,max_product_id=110,num_rows = 5000000):
    
    synthetic_data = pd.DataFrame({
        "account_id": np.random.randint(1000000, 2000000, num_rows),
        "product_id": np.random.randint(min_product_id, max_product_id, num_rows),
        "txn_amt": np.random.uniform(10, 1000, num_rows).round(2),
        # "txn_flow": np.random.choice(["credit", "debit","valuechain"], num_rows),
        "date_id": np.random.randint(6000, 6730, num_rows),
    })
    synthetic_data["txn_flow"] = txn_flow
    return synthetic_data
```

```python
debit = generate_synthetic_data("debit",min_product_id=10,max_product_id=110,num_rows = 5000000)
```

```python
credit = generate_synthetic_data("credit",min_product_id=10,max_product_id=110,num_rows = 3000000)
```

```python
service = generate_synthetic_data("services",min_product_id=111,max_product_id=211,num_rows = 2000000)
```

```python
all_df = pd.concat([debit,credit,service])
all_df
```

```python
all_df['account_id'].nunique()
```

```python
all_df.drop_duplicates(subset=["account_id","product_id","date_id","txn_flow"])
```

```python
debit2 = generate_synthetic_data("debit",min_product_id=10,max_product_id=110,num_rows = 265)
```

```python
all_df = pd.concat([all_df,debit2])
all_df
```

```python
all_df.drop_duplicates(subset=["account_id","product_id","date_id","txn_flow"],inplace=True)
```

```python
# Save the synthetic data to a CSV file
synthetic_file_path = "synthetic_customer_churn_data.csv"
all_df.to_csv(synthetic_file_path, index=False,)

synthetic_file_path
```
